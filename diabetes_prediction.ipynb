```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn import linear_model
import sklearn.metrics as sm
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

df = pd.read_csv('diabetes2.csv')

sns.countplot (x= 'Outcome', data= df)

plt.subplots(figsize = (8, 8))
sns.heatmap(df.corr(), annot= True)
plt.show()
```

```python
x = df.drop(["Outcome"], axis=1) 
y = df.Outcome

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

```python
reg_linear = linear_model.LogisticRegression()
reg_linear.fit(x_train, y_train)

out_pred = reg_linear.predict(x_test)
```

```python
out_pred
```

```python
y_test = np.array(y_test)
```

```python
cor = 0
incor = 0
for i in range(y_test.size):
    if out_pred[i] == np.array(y_test[i]):
        cor += 1
    elif out_pred[i] !=  np.array(y_test[i]):
        incor += 1
```

```python
print("sm.mean_squared_error= ", sm.mean_squared_error(y_test, out_pred))
```

```python
err = np.abs(y_test - out_pred)
err.mean()
```
